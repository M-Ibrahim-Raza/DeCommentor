### **Commands** 

#### **Command-Line Options** 

| **Option**         | **Argument Type**     | **Description**                                                                                                                    | **Example Usage**                                                |
| ------------------ | --------------------- | ---------------------------------------------------------------------------------------------------------------------------------- | ---------------------------------------------------------------- |
| `path`             | `<file or directory>` | Specifies the file or directory to process. This is a **required** positional argument.                                            | `python script.py myscript.py`<br>`python script.py my_project/` |
| `-o`, `--output`   | `<directory>`         | Defines an **output directory** where cleaned files will be saved. If not provided, creates `.cleaned.py` files next to originals. | `python script.py src/ -o cleaned_src/`                          |
| `-i`, `--in-place` | *(flag)*              | Cleans files **in-place**, overwriting the originals. Cannot be used with `--output`.                                              | `python script.py src/ --in-place`                               |
| `--ignore`         | `<dir1> <dir2> ...`   | List of directory names to **skip** during processing (default: `venv __pycache__ .git .tox node_modules .pytest_cache`).          | `python script.py src/ --ignore tests venv`                      |
| `--ignore-files`   | `<file1> <file2> ...` | List of file names to **skip** during processing.                                                                                  | `python script.py src/ --ignore-files setup.py conftest.py`      |

#### **Commands** 

| **Command**                                                                  | **Description**                                 | **Effect / Output**                                                                                                 |
| ---------------------------------------------------------------------------- | ----------------------------------------------- | ------------------------------------------------------------------------------------------------------------------- |
| `python script.py input.py`                                                  | Process a **single Python file**                | Creates a new file named `input.cleaned.py` in the same directory                                                   |
| `python script.py project/ -o cleaned_project/`                              | Process an **entire directory**                 | Cleans all `.py` files in `project/` and saves cleaned versions in `cleaned_project/` (preserving folder structure) |
| `python script.py project/ --in-place`                                       | Process all files **in-place**                  | Directly modifies all `.py` files in `project/` (overwrites originals)                                              |
| `python script.py project/ --ignore tests venv __pycache__`                  | Process directory but **skip specific folders** | Ignores any subfolders named `tests`, `venv`, or `__pycache__` during cleaning                                      |
| `python script.py project/ --ignore-files setup.py conftest.py`              | Process directory but **skip specific files**   | Ignores files named `setup.py` and `conftest.py` during cleaning                                                    |
